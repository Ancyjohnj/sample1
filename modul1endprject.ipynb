```python
import pandas as pd 
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_excel('ABC Company (6).xlsx')
print(df.head())
```

```python
df.shape
```

```python
df.columns
```

```python
# height column preprocessing
```

```python
df["Height"]=np.random.randint(150,181,size=len(df))
df["Height"].head()
```

```python
#team_wise employee count
team_counts=df["Team"].value_counts()
print(team_counts)
```

```python
#team percentage
team_percentage=(df["Team"].value_counts()/len(df))*100
print(team_percentage)
```

```python
#bar graph
team_counts.plot(kind="bar",figsize=(10,5))
plt.title("Employees in Each Team")
plt.xlabel("Team")
plt.ylabel("Number of Employees")
plt.show()
```

```python
#position distribution
position_count=df["Position"].value_counts()
print(position_count)
```

```python
#bar gragh
position_count.plot(kind="bar")
plt.title("employee by positions")
plt.xlabel("Position")
plt.ylabel("number of employees")
plt.show()
```

```python
#predominant age group
bins=[20,30,40,50,60]
labels=["20-30","30-40","40-50","50-60"]
df["Age_Group"]=pd.cut(df["Age"],bins=bins,labels=labels)
print(df["Age_Group"].value_counts())
```

```python
#graph
df["Age_Group"].value_counts().plot(kind="bar")
plt.title("employees by age group")
plt.xlabel("age group")
plt.ylabel("number of empolyees")
plt.show()
```

```python
#high salary expenditure
salary_team=df.groupby("Team")["Salary"].sum()
print(salary_team)
```

```python
#graph
salary_team.plot(kind="bar")
plt.title("total salary expenditurw by team")
plt.xlabel("team")
plt.ylabel("total salary")
plt.show()
```

```python
#age vs salary correlation
correlation=df["Age"].corr(df["Salary"])
print("Correlation:",correlation)
```

```python
#scatter plot
plt.scatter(df["Age"],df["Salary"])
plt.title("age vs salary")
plt.xlabel("age")
plt.ylabel("salary")
plt.show()
```
