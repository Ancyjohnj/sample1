```python
import seaborn as sns
import pandas as pd
import matplotlib.pyplot as plt
df=sns.load_dataset("titanic")
df.head()
```

```python
#1,count:passenger count by embarked port
.create a countplot to display the number of passengers who boarded from each port(embarked),separated by sex.
```

```python
plt.figure(figsize=(8,5))
sns.countplot(x='embarked',hue='sex',data=df)
plt.title('Passenger Count by Embarked Port and Gender')
plt.xlabel('Embarked Port-C:Cherbourg,Q:Queenstown,S:Southampton')
plt.ylabel('Passenger Count')
plt.legend(title='sex')
plt.show()
```

```python
#2,swarmplot:age distribution across classes
.create a swarmplot to visualize the distribution of age across passenger classes(pclass),categorized by sex.
```

```python
plt.figure(figsize=(9,6))
sns.swarmplot(x='pclass',y='age',hue='sex',data=df)
plt.title('Age Distribution across Passenger Classes by Gender')
plt.xlabel('Passenger Class')
plt.ylabel('Age')
plt.legend(title='sex')
plt.show()              
```

```python
#3,pie chart:gender disribution
.create a pie chart using mtaplotlib to show the percentage of male vs.female passengers.-add percentage labels using autopct-use explode for one slice to highlight it.
```

```python
df['sex'].value_counts().plot(kind='pie',autopct='%1.f%%',colors=['orange','blue'])
plt.title('Gender Distribution')
plt.ylabel('')
plt.legend(['Male','Female'])
plt.show()
```

```python
#4,heatmap:correlation matrix
.create a heatmap to show correlations between age,fare,pclass,sibsp,parch,and survived.(use seabron .heatmap with annotations)
```

```python
plt.figure(figsize=(8,6))
sns.heatmap(df[['fare','pclass','sibsp','parch','age','survived']].corr(),annot=True,cmap='coolwarm')
plt.title('Correlation Heatmap')
plt.show()
```

```python
#5,violin plot:fare by class and gender
.create a violin plot to visualize fare distribution across classes,separated by gender.
```

```python
sns.violinplot(x='pclass',y='fare',hue='sex',data=df)
plt.title('Fare Distribution by Class andGender')
plt.xlabel('Passenger Class')
plt.ylabel('Fare')
plt.legend(title='sex')
plt.show()
```

```python
#6,scatter plot:age vs fare
.create a scatter plot to explore the relationship between age and fire,colored by surival status.
.use a seabron scatterplot with hue='survived'.each plot must have titles,labels,and legends where needed.
```

```python
sns.scatterplot(x='age',y='fare',hue='survived',data=df)
plt.title('Age vs Fare colored by Survival Status')
plt.xlabel('Age')
plt.ylabel('Fare')
plt.legend(title='Survived',labels=['No','Yes'])
plt.show()
```
